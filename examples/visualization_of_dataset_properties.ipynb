### Import the necessary packages

```python
from belearn.dataset.dataset import BE_Dataset
from m3util.util.IO import download_and_unzip
import matplotlib.pyplot as plt
import numpy as np
```

### Load the h5 file

```python
# Download the data file from Zenodo
url = 'https://zenodo.org/record/7774788/files/PZT_2080_raw_data.h5?download=1'

# Specify the filename and the path to save the file
# filename = 'data_raw.h5'

filename = 'data_raw.h5'
save_path = './Data'

# download the file
download_and_unzip(filename, url, save_path)
```

```python
data_path = save_path + '/' + filename

# instantiate the dataset object
dataset = BE_Dataset(data_path)

# print the contents of the file
dataset.print_be_tree
```

### Plot of properties that are singular numbers

```python
print("number of pixels: ", dataset.num_pix, "\nThis is the total number of pixels in the dataset\n")
print("number of bins: ", dataset.num_bins, "\nThis is the number of bins in the dataset\n")
print("BE center frequency: ", dataset.be_center_frequency, "\nThis is the center frequency of the BE waveform\n")
print("BE bandwidth: ", dataset.be_bandwidth, "\nThis is the bandwidth of the BE waveform\n")
print("BE repeats: ", dataset.be_repeats, "\nThis is the number of times the BE waveform is repeated\n")
print("number of cycles: ", dataset.num_cycles, "\nThis is the number of cycles in the dataset\n")
print("voltage steps: ", dataset.voltage_steps, "\nThis is the number of voltage steps in the dataset\n")
print("spectroscopic length (num_bins * voltage_steps): ", dataset.spectroscopic_length)
print("Sampling rate: ", format(dataset.sampling_rate, ".3e"), "\nThis is the sampling rate of the dataset\n")
```

### Plot of BE bandwidth
